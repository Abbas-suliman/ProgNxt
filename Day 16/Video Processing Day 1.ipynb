## basic Feature

```python
import cv2
cap = cv2.VideoCapture("slow_traffic_small.mp4")

while True:
    ret, frame = cap.read() 
    resized = cv2.resize(frame, (800,480))
    if not ret:
        break
    Grayed = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)
    Blurredd = cv2.GaussianBlur(Grayed, (5,5),0)
    edges = cv2.Canny(Blurredd, 50, 150)
    cv2.imshow("Original video", resized)
    cv2.imshow("edge detection video", edges)
    if cv2.waitKey(25) & 0xFF ==ord("q"):
        break

cap.release() 
cv2.destroyAllWindows() 
```

## sobel Edge detection 

```python
import cv2
cap = cv2.VideoCapture("slow_traffic_small.mp4")

while True:
    ret, frame = cap.read() 
    resized = cv2.resize(frame, (800,480))
    if not ret:
        break
    Grayed = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)

    # sobel X and Y
    sobel_x = cv2.Sobel(Grayed, cv2.CV_64F, 1,0, ksize = 3)
    sobel_y = cv2.Sobel(Grayed, cv2.CV_64F, 0,1, ksize = 3)

    # combine X and Y
    sobel = cv2.magnitude(sobel_x,sobel_y)
    sobel = cv2.convertScaleAbs(sobel)

    
    cv2.imshow("Original video", resized)
    cv2.imshow("edge detection video", sobel)
    if cv2.waitKey(25) & 0xFF ==ord("q"):
        break

cap.release() 
cv2.destroyAllWindows() 
```

## Laplacian

```python
import cv2
cap = cv2.VideoCapture("slow_traffic_small.mp4")

while True:
    ret, frame = cap.read() 
    resized = cv2.resize(frame, (800,480))
    if not ret:
        break
    Grayed = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)

       # Laplacian
    Laplacian = cv2.Laplacian(Grayed,cv2.CV_64F)
    Laplacian = cv2.convertScaleAbs(Laplacian)

    
    cv2.imshow("Original video", resized)
    cv2.imshow("edge detection video", Laplacian)
    if cv2.waitKey(25) & 0xFF ==ord("q"):
        break

cap.release() 
cv2.destroyAllWindows() 
```
