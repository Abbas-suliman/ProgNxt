```python
from ultralytics import YOLO
import cv2
import time
MODEL_NAME = "yolo11n.pt"
INPUT_FILE = "slow_traffic_small.mp4"
OUTPUT_FILE = "yolo_detection_workflow_output.mp4"
CONFIDENCE = 0.40
IOU_THRESHOLD = 0.50
model = YOLO(MODEL_NAME)
capture = cv2.VideoCapture(INPUT_FILE)
if not capture.isOpened():
    raise FileNotFoundError(f"Unable to open {INPUT_FILE}")
fps = capture.get(cv2.CAP_PROP_FPS)
width = int(capture.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(capture.get(cv2.CAP_PROP_FRAME_HEIGHT))
if fps <= 0:
    fps = 25.0
writer = cv2.VideoWriter( OUTPUT_FILE,    cv2.VideoWriter_fourcc(*"mp4v"),    fps,    (width, height)    )
frame_number = 0
total_detections = 0
while True:
    success, frame = capture.read()
    if not success:
        break
    frame_number += 1
    start = time.perf_counter()
    result = model(frame, conf=CONFIDENCE,iou=IOU_THRESHOLD, verbose=False )[0]
    elapsed = time.perf_counter() - start
    annotated = result.plot()
    count = 0 if result.boxes is None else len(result.boxes)
    total_detections += count
    fps_text = 1.0 / elapsed if elapsed > 0 else 0.0
    cv2.putText( annotated,f"Frame: {frame_number}",(15, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2 )
    cv2.putText( annotated,  f"Objects: {count}",  (15, 60),cv2.FONT_HERSHEY_SIMPLEX,0.7, (0, 255, 0), 2  )
    cv2.putText(  annotated, f"Inference FPS: {fps_text:.1f}",(15, 90),cv2.FONT_HERSHEY_SIMPLEX,0.7, (0, 255, 0), 2 )
    writer.write(annotated)
    cv2.imshow("YOLO Detection Workflow Application", annotated)
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break
capture.release()
writer.release()
cv2.destroyAllWindows()
print("Frames processed:", frame_number)
print("Total frame-level detections:", total_detections)
print("Saved output:", OUTPUT_FILE)
```
