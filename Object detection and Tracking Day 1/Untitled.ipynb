```python
from ultralytics import YOLO
import cv2
MODEL_NAME = "yolo11n.pt"
INPUT_FILE = "shapes.jpg"
CONFIDENCE = 0.40
model = YOLO(MODEL_NAME)
results = model(INPUT_FILE,conf=CONFIDENCE,verbose=False)
result = results[0]
annotated = result.plot()
print("Topic: Image Classification vs Object Detection")
print("Detected boxes:", 0 if result.boxes is None else len(result.boxes))
if result.boxes is not None:
    for box in result.boxes:
        class_id = int(box.cls[0])
        confidence = float(box.conf[0])
        print("Class:",result.names[class_id],"Confidence:",round(confidence, 3))
cv2.imshow("Detection Result", annotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
