```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def display(img, cmap = None):
    fig = plt.figure(figsize=(10,10))
    ax = fig.add_subplot(111)
    ax.imshow(img, cmap = cmap)
```

## color modes

```python
img = cv2.imread('image/burano.jpg')
display(img)
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
display(img_rgb)
```

```python
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
display(img_gray,cmap = 'gray')
```

```python
fig, axs = plt.subplots(nrows = 1, ncols = 3, figsize = (20,20))
for i in range(0,3):
    ax = axs[i]
    ax.imshow(img_rgb[:,:,i], cmap = 'gray')
plt.show()
```

```python
img_HSV = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
img_HSL = cv2.cvtColor(img, cv2.COLOR_BGR2HLS)
fig, (ax1, ax2) = plt.subplots(nrows = 1, ncols = 2, figsize = (20,20))
ax1.imshow(img_HSV)
ax2.imshow(img_HSL)
plt.show()
```

## Draw on Image

```python
img2 = cv2.imread('image/wall.jpg')
img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
display(img2)
```

```python
img2_copy = img2.copy()
# Draw a rectangle
cv2.rectangle(img2_copy, pt1 = (100,200), pt2 = (200,400), color = (255,0,0), thickness = 4)
display(img2_copy)
```

```python
# draw a circle
cv2.circle(img2_copy, center = (600,410), radius = 50, color = (0,255,0), thickness = 5)
display(img2_copy)
```

```python
# Draw a rectangle
cv2.rectangle(img2_copy, pt1 = (780,380), pt2 = (820,420), color = (255,0,0), thickness = 4)
display(img2_copy)
```

```python
# text
cv2.putText(img2_copy, text = "The Wall of Love", org = (250,250), fontFace = cv2.FONT_HERSHEY_PLAIN, 
            fontScale = 2, color = (0,0,255), thickness = 2)
display(img2_copy)
```
