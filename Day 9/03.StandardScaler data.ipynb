```python
from sklearn import preprocessing
import numpy as np
```

```python
from sklearn import datasets
iris = datasets.load_iris()
```

```python
X = iris.data[:, [1,2]]
Y = iris.target

print(X)
```

```python
std_scale = preprocessing.StandardScaler().fit(X)
X_std = std_scale.transform(X)

minmax_scale = preprocessing.MinMaxScaler().fit(X)
X_minmax = minmax_scale.transform(X)


print('mean before Stadardization: petal length (cm) = {:.1f},sepal width (cm) = {:.1f}'
    .format(X[:,0].mean(),X[:,1].mean()))


print('SD before Stadardization: petal length (cm) = {:.1f},sepal width (cm) = {:.1f}'
    .format(X[:,0].std(),X[:,1].std()))
```

```python
print('mean after Stadardization: petal length (cm) = {:.1f},sepal width (cm) = {:.1f}'
    .format(X_std[:,0].mean(),X_std[:,1].mean()))


print('SD after Stadardization: petal length (cm) = {:.1f},sepal width (cm) = {:.1f}'
    .format(X_std[:,0].std(),X_std[:,1].std()))
```

```python
print('mean after minimax: petal length (cm) = {:.1f},sepal width (cm) = {:.1f}'
    .format(X_minmax[:,0].mean(),X_minmax[:,1].mean()))


print('SD after minimax: petal length (cm) = {:.1f},sepal width (cm) = {:.1f}'
    .format(X_minmax[:,0].std(),X_minmax[:,1].std()))
```
