```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(-5,5,0.01)
```

```python
def plot(func,yaxis = (-1.5, 1.5)):
    plt.ylim(yaxis)
    plt.locator_params(nbins = 5)
    plt.xticks(fontsize = 14)
    plt.yticks(fontsize = 14)
    plt.axhline(lw = 1, c = 'black')
    plt.axvline(lw = 1, c = 'black')
    plt.box(on = None)
    plt.grid(alpha = 0.4, ls = '-.')
    plt.plot(x, func(x), c='r', lw=3)
```

```python
# Binary

binary_step = np.vectorize(lambda x:1 if x>0 else 0, otypes = [np.float16])

plot(binary_step,yaxis = (-1.5, 1.5))
```

```python
# bipolar
bipolar_step = np.vectorize(lambda x:1 if x>0 else -1, otypes = [np.float16])

plot(bipolar_step,yaxis = (-1.5, 1.5))
```

```python
# sigmoid
def sig(x):
    return 1/(1+np.exp(-x))


plot(sig,yaxis = (-0.4, 1.4)) # changed the axis
```

```python
# TanH
def TanH(x):
    return 2/(1+np.exp(-2*x))-1


plot(TanH) # default y axis
```

```python
# Relu
ReLu = np.vectorize(lambda x:x if x>0 else 0, otypes = [np.float16])

plot(ReLu,yaxis = (-0.4, 1.4)) # check with the axis 
```

```python
# LeakyRelu
#LeakyRelu = np.vectorize(lambda x:max(0.1*x,x), otypes = [np.float16]) # according to the concept
LeakyRelu = np.vectorize(lambda x:x if x>=0 else 0.1*x, otypes = [np.float16]) # accordgin to the equation 
plot(LeakyRelu,yaxis = (-0.4, 1.4)) # check with the axis 
```

```python
# Elu
Elu = np.vectorize(lambda x:x if x>0 else 0.1*(np.exp(x)-1), otypes = [np.float16])

plot(Elu,yaxis = (-0.4, 1.4)) # check with the axis 
```

```python
# softplus
def softplus(x):
    return np.log(1+np.exp(x))

plot(softplus,yaxis = (-0.4, 1.4))
```
