```python
# Cartooooooonn
import cv2
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break
    color = cv2.bilateralFilter(frame, 9, 75,75)
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    gray = cv2.medianBlur(gray, 7)
    edges = cv2.adaptiveThreshold(gray, 255,cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 9, 2)
    cartoon = cv2.bitwise_and(color, color, mask = edges)
    
    cv2.imshow("Cartoon", cartoon)  
    if cv2.waitKey(1)&0xFF ==27:
        break
cap.release()
cv2.destroyAllWindows()
```

```python
# Animeeee

import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    # Smooth image
    smooth = frame.copy()

    for i in range(2):
        smooth = cv2.bilateralFilter(
            smooth,
            9,
            100,
            100
        )

    # Convert to grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Reduce noise
    gray = cv2.medianBlur(gray, 5)

    # Detect edges
    edges = cv2.adaptiveThreshold(
        gray,
        255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        9,
        5
    )

    # Combine
    cartoon = cv2.bitwise_and(
        smooth,
        smooth,
        mask=edges
    )

    cv2.imshow("Realtime Cartoon", cartoon)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2

cap = cv2.VideoCapture(0)

face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    # Flip camera like a mirror
    frame = cv2.flip(frame, 1)

    # Cartoon effect
    color = cv2.bilateralFilter(frame, 9, 100, 100)

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    gray = cv2.medianBlur(gray, 5)

    edges = cv2.adaptiveThreshold(
        gray,
        255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        9,
        3
    )

    cartoon = cv2.bitwise_and(color, color, mask=edges)

    # Detect face
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(100, 100)
    )

    for (x, y, w, h) in faces:

        # Avatar border
        cv2.ellipse(
            cartoon,
            (x + w // 2, y + h // 2),
            (w // 2, h // 2),
            0,
            0,
            360,
            (0, 255, 255),
            3
        )

        # Avatar label
        cv2.putText(
            cartoon,
            "AVATAR",
            (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (0, 255, 255),
            2
        )

    cv2.imshow("My Cartoon Avatar", cartoon)

    # Press Q to exit
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
