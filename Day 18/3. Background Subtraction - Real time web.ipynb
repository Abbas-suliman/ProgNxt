```python
# Background Subtraction  - MOG2

import cv2
cap = cv2.VideoCapture(0)
background = cv2.createBackgroundSubtractorMOG2(history=500,varThreshold=25,detectShadows=True)

while True:
    ret, frame = cap.read()
    if not ret:
        break
    mask = background.apply(frame)
    cv2.imshow("Original", frame)
    cv2.imshow("Foreground Mask ", mask)
    if cv2.waitKey(1)&0xFF ==27:
        break

cap.release()
cv2.destroyAllWindows()
```

```python
# foreground objects

import cv2
cap = cv2.VideoCapture(0)
background = cv2.createBackgroundSubtractorMOG2(history=500,varThreshold=40,detectShadows=False)
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5,5))

while True:
    ret, frame = cap.read()
    if not ret:
        break
    mask = background.apply(frame)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
    mask = cv2.dilate(mask, kernel, iterations=2)
    countors, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    for c in countors:
        area = cv2.contourArea(c)
        if area<1000:
            continue
        x,y,w,h = cv2.boundingRect(c)
        cv2.rectangle(frame, (x, y), (x+w,y+h), (0,0,255),2)
        cv2.putText(frame,f"Area{int(area)}",(x,y-8),cv2.FONT_HERSHEY_SIMPLEX,.5,(0, 255, 0), 2
    )

    
    cv2.imshow("Original", frame)
    cv2.imshow("KNN Foreground Mask ", mask)
    if cv2.waitKey(1)&0xFF ==27:
        break

cap.release()
cv2.destroyAllWindows()
```
