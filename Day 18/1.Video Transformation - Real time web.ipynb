## Video Transformation

```python
# resize live camera

import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break
    resized = cv2.resize(frame, (640,480))
    cv2.imshow("Realtime resize", resized)
    if cv2.waitKey(1)&0xFF ==27:
        break
    
cap.release()
cv2.destroyAllWindows()
```

```python
# dynamic resize

import cv2

cap = cv2.VideoCapture(0)
scale = 1.0

while True:
    ret, frame = cap.read()
    if not ret:
        break
    height, width = frame.shape[:2]
    out = cv2.resize(frame, (max(1,int(width*scale)),max(1,int(height*scale))))
    cv2.imshow("Dynamic resize", out)
    key = cv2.waitKey(1)&0xFF
    if  key==ord('q'):
        break
    elif key==ord('=') or key==ord('+'):
        scale = min(2.0, scale+0.1)
    elif key==ord('_') or key==ord('-'):
        scale = max(0.2, scale-0.1)    
cap.release()
cv2.destroyAllWindows()
```

```python
# ROI_Pipeline on live cam

import cv2
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break
    height, width = frame.shape[:2]
    x1, y1 = width//4, height//4
    x2, y2 = 3*width//4, 3*height//4

    roi = frame[y1:y2, x1:x2]
    roi = cv2.resize(roi, None, fx = 1.2, fy=1.2)
    roi = cv2.resize(roi, (x2-x1, y2-y1))
    roi = cv2.GaussianBlur(roi, (9,9),0)

    output = frame.copy()
    output[y1:y2, x1:x2] = roi
    cv2.rectangle(output, (x1, y1), (x2,y2), (255,255,255),2)

    
    cv2.imshow("ROI transformation", output)
    if cv2.waitKey(1)&0xFF ==27:
        break
cap.release()
cv2.destroyAllWindows()
```
