```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def display(img, cmap = None):
    fig = plt.figure(figsize=(10,10))
    ax = fig.add_subplot(111)
    ax.imshow(img, cmap = cmap)
```

```python
img  = cv2.imread('simpson.jpg')
img  = cv2.bitwise_not(img)
display(img)
```

```python
# Erosion

ker_0  = np.ones((9,9),np.uint8)
ker_1 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (9,9))
ker_2 = cv2.getStructuringElement(cv2.MORPH_CROSS, (9,9))

kernels = [ker_0,ker_1,ker_2]

plt.figure(figsize = (20,20))
for i in range(3):
    img_copy = img.copy()
    img_copy = cv2.erode(img_copy, kernels[i],iterations = 3)
    plt.subplot(1,3,i+1)
    plt.imshow(img_copy)
    plt.axis('off')
plt.show()
```

```python
ker_0  = np.ones((7,7),np.uint8)
ker_1 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (7,7))
ker_2 = cv2.getStructuringElement(cv2.MORPH_CROSS, (7,7))
kernels = [ker_0,ker_1,ker_2]
plt.figure(figsize = (20,20))
for i in range(3):
    img_copy = img.copy()
    img_copy = cv2.erode(img_copy, kernels[i],iterations = 3)
    plt.subplot(1,3,i+1)
    plt.imshow(img_copy)
    plt.axis('off')
plt.show()
```

```python
# dilation
ker_0  = np.ones((7,7),np.uint8)
ker_1 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (7,7))
ker_2 = cv2.getStructuringElement(cv2.MORPH_CROSS, (7,7))
kernels = [ker_0,ker_1,ker_2]
plt.figure(figsize = (20,20))
for i in range(3):
    img_copy = img.copy()
    img_copy = cv2.dilate(img_copy, kernels[i],iterations = 3)
    plt.subplot(1,3,i+1)
    plt.imshow(img_copy)
    plt.axis('off')
plt.show()
```

```python
ker_0  = np.ones((7,7),np.uint8)

img_open = cv2.morphologyEx(img, cv2.MORPH_OPEN, ker_0)
img_close = cv2.morphologyEx(img, cv2.MORPH_CLOSE, ker_0)
img_grad = cv2.morphologyEx(img, cv2.MORPH_GRADIENT, ker_0)
img_tophat = cv2.morphologyEx(img, cv2.MORPH_TOPHAT, ker_0)
img_blackhat = cv2.morphologyEx(img, cv2.MORPH_BLACKHAT, ker_0)

images = [img, img_open,img_close, img_grad, img_tophat, img_blackhat]

plt.figure(figsize = (20,10))
for i in range(6):
    plt.subplot(2,3,i+1)
    plt.imshow(images[i], cmap ='gray')
    plt.axis('off')

plt.tight_layout()
plt.show()
```
