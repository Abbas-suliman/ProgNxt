```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
img_1 = cv2.imread('image1.jpeg')
img_2 = cv2.imread('image2.jpeg')
```

```python
def display(img, cmap = None):
    fig = plt.figure(figsize=(10,10))
    ax = fig.add_subplot(111)
    ax.imshow(img, cmap = cmap)
```

```python
display(img_1,cmap = None)
```

```python
display(img_2,cmap = None)
```

```python
# AND, OR, NOT
# cv2.bitwise_<>(im1, im2, dest, mask=none)

dest_and = cv2.bitwise_and(img_1, img_2, mask = None)
cv2.imshow("Bitwise_AND", dest_and)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# ex 2
dest_and = cv2.bitwise_and(img_2, img_1, mask = None)
cv2.imshow("Bitwise_AND", dest_and)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
dest_or = cv2.bitwise_or(img_2, img_1, mask = None)
cv2.imshow("Bitwise_OR", dest_or)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
dest_not = cv2.bitwise_not(img_2, mask = None)
cv2.imshow("Bitwise_NOT", dest_not)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
dest_xor = cv2.bitwise_xor(img_2,img_1, mask = None)
cv2.imshow("Bitwise_XOR", dest_xor)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
