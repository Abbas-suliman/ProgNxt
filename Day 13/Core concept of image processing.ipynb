## 1. Image information

```python
import cv2
img = cv2.imread('messi5.jpg')
print("Shape:",img.shape)
print("data type:" ,img.dtype)
print("Height and Width: ",img.shape[:2])
```

```python
# Ex 2
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print("gray Shape:",gray.shape)
print("max/ min: ", gray.min(), gray.max())
```

## 2. Image Normalization

```python
# pixel intensity to numeric range
# ML optim, feature extraction
img = cv2.imread('messi5.jpg')
norm = cv2.normalize(img, None, 0,1, cv2.NORM_MINMAX, cv2.CV_32F)

print(norm.min(), norm.max())
```

```python
cv2.imwrite("norm_new.png", norm)
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

norm8 = cv2.normalize(gray, None, 0, 255, cv2.NORM_MINMAX)

cv2.imwrite("normalized.png", norm8)
```

```python
gr = cv2.imread('normalized.png')
print(gr.min(), gr.max())
```

## 3. Negative transformation

```python
img3 = cv2.imread('messi5.jpg')
negative = 255-img3
cv2.imwrite("neg_img.jpg", negative)
```

```python
gry3 = cv2.cvtColor(img3, cv2.COLOR_BGR2GRAY)
negative = cv2.bitwise_not(gry3)
cv2.imwrite("neg_img_3.jpg", negative)
```

## 4. contrast stretching


```python
import numpy as np
gry4 = cv2.cvtColor(img3, cv2.COLOR_BGR2GRAY)
lo, hi = gry4.min(), gry4.max()
stretch = ((gry4-lo)*255.0/(hi-lo)).clip(0,255).astype(np.uint8)

cv2.imwrite("stretchedddd.jpg", stretch)
```

```python
# percentile

p2, p98 = np.percentile(gry4, (2,98))
stretch_1 = np.clip((gry4-p2)*255/(p98-p2), 0, 255).astype(np.uint8)

cv2.imwrite("stretched_1.jpg", stretch)
```

```python
p5, p95 = np.percentile(gry4, (5,95))
stretch_1 = np.clip((gry4-p5)*255/(p95-p5), 0, 255).astype(np.uint8)

cv2.imwrite("stretched_2.jpg", stretch)
```

## 6. CLAHE

```python
gry7 = cv2.cvtColor(img6, cv2.COLOR_BGR2GRAY)
clahe = cv2.createCLAHE(clipLimit = 2.0, tileGridSize = (8,8))
out = clahe.apply(gry7)
cv2.imwrite("clahe_img.jpg", out)
```

## 7. salt and pepper noise

```python
img6 = cv2.imread('outputs_document_scan.jpg')
gry7 = cv2.cvtColor(img6, cv2.COLOR_BGR2GRAY)

noisy = gry7.copy()

p =np.random.rand(*gry7.shape)
noisy[p<0.02]  =0
noisy[p>0.98] = 255

clean = cv2.medianBlur(noisy,5)

cv2.imwrite("Salt.jpg", clean)
```

```python
mean = cv2.blur(noisy,(5,5))
cv2.imwrite("mean_blur.jpg", mean)
```

```python
gauss = cv2.GaussianBlur(noisy,(5,5),0)
cv2.imwrite("gauss_blur.jpg", gauss)
```
