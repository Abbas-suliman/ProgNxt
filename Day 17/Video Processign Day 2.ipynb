## Processing the frame

```python
cap = cv2.VideoCapture("running.mp4")
frame_no = 0
while True:
    ret, frame = cap.read()
    if not ret:
        break
    frame_no +=1
    # print("Processing the frame: ", frame_no)
    #cv2.putText(frame, "Frame Processing", (30,50), cv2.FONT_HERSHEY_SIMPLEX, 1, (255,0,0),2)
    cv2.putText(frame, f"Frame: {frame_no}", (200,140), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,255,0), 2)
    cv2.imshow("Processed video", frame)
    if cv2.waitKey(1)&0xFF ==27:
        break
    
cap.release()
cv2.destroyAllWindows()
```

## split the video color channel

```python
# split the video color channel
cap = cv2.VideoCapture("running.mp4")

while True:
    ret, frame = cap.read()
    if not ret:
        break
    b,g,r = cv2.split(frame)
    
    cv2.imshow("Blue", b)
    cv2.imshow("Green", g)
    cv2.imshow("Red", r)
    if cv2.waitKey(1)&0xFF ==27:
        break
    
cap.release()
cv2.destroyAllWindows()
```

## reorder the channel

```python
# reorder the channel

cap = cv2.VideoCapture("running.mp4")

while True:
    ret, frame = cap.read()
    if not ret:
        break
    b,g,r = cv2.split(frame)
    merged = cv2.merge([b,g,r])
    # cv2.imshow("Blue", b)
    # cv2.imshow("Green", g)
    # cv2.imshow("Red", r)
    cv2.imshow("Reorderd Channel", merged)
    if cv2.waitKey(1)&0xFF ==27:
        break
    
cap.release()
cv2.destroyAllWindows()
```

## control playback speed

```python
# control playback speed

cap = cv2.VideoCapture("rain.mp4")

fps = cap.get(cv2.CAP_PROP_FPS)
delay = max(1,int(1000/fps))

while True:
    ret, frame = cap.read()
    if not ret:
        break
    cv2.imshow("normal Speed", frame)
    if cv2.waitKey(delay)&0xFF ==27:
        break
    
cap.release()
cv2.destroyAllWindows()
```

## blur selected area / ROI

```python
# blur selected area / ROI
cap = cv2.VideoCapture("slow_traffic_small.mp4")

while True:
    ret, frame = cap.read()
    if not ret:
        break
    roi = frame[100:300,200:500]
    blurred = cv2.GaussianBlur(roi,(31,31),0)
    frame[100:300,200:500] = blurred
    cv2.imshow("blur selected area", frame)
    if cv2.waitKey(1)&0xFF ==27:
        break
    
cap.release()
cv2.destroyAllWindows()
```

## blur center area / ROI

```python
# blur center area / ROI
cap = cv2.VideoCapture("slow_traffic_small.mp4")

while True:
    ret, frame = cap.read()
    if not ret:
        break
    height, width = frame.shape[:2]
    
    roi = frame[height//4:height//2,width//4:3*width//4]
    blurred = cv2.GaussianBlur(roi,(31,31),0)
    frame[height//4:height//2,width//4:3*width//4] = blurred
    cv2.imshow("blur center area", frame)
    if cv2.waitKey(1)&0xFF ==27:
        break
    
cap.release()
cv2.destroyAllWindows()
```

## Threshold

```python
# Threshold
import cv2
cap = cv2.VideoCapture("running.mp4")

while True:
    ret, frame = cap.read()
    if not ret:
        break
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    _, binary = cv2.threshold(gray,120,255,cv2.THRESH_BINARY)
    #print("Threshold the video: ", binary)
    cv2.imshow("Threshold the video: ", binary)
    if cv2.waitKey(1)&0xFF ==27:
        break

cap.release()
cv2.destroyAllWindows()
```

##  Adaptive Threshold

```python
# Adaptive Threshold
import cv2
cap = cv2.VideoCapture("running.mp4")

while True:
    ret, frame = cap.read()
    if not ret:
        break
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    binary = cv2.adaptiveThreshold(gray,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY,11,2)
    
    cv2.imshow("Adaptive Threshold the video: ", binary)
    if cv2.waitKey(1)&0xFF ==27:
        break

cap.release()
cv2.destroyAllWindows()
```

## recored web cam video

```python
import cv2
cap = cv2.VideoCapture(0)

fps = 20.0
fram_hei = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fram_widt = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))

out = cv2.VideoWriter("Webcam.mp4",cv2.VideoWriter_fourcc(*"mp4v"),fps,(fram_widt,fram_hei))
while True:
    ret, frame = cap.read()
    if not ret:
        break
    out.write(frame)
    cv2.imshow("Recording ", frame)
    if cv2.waitKey(1)&0xFF ==27:
        break

cap.release()
out.release()
cv2.destroyAllWindows()
```

## frame extractor => specific time stamp

```python
import cv2
sec = float(input("enter the timestamp in sec: "))

cap = cv2.VideoCapture("slow_traffic_small.mp4")

cap.set(cv2.CAP_PROP_POS_MSEC, sec * 1000)
ret, frame = cap.read()

if not ret:
    raise SystemExit("unable to read requested frame")

filename = f"frame_{sec:.2f}s.jpg"
cv2.imwrite(filename, frame)

print("Saved: ", filename)
cap.release()
```

## Multi time => snapshot generator

```python
import cv2
timestamp = [0,14,27,9,15,20,26]

cap = cv2.VideoCapture("slow_traffic_small.mp4")
for sec in timestamp:
    cap.set(cv2.CAP_PROP_POS_MSEC, sec * 1000)
    ret, frame = cap.read()
    if not ret:
        #######raise SystemExit("unable to read requested frame")
        print("Failed: ",sec, "seconds")
        continue
    filename = f"snapshot_{sec:.2f}s.jpg"
    cv2.imwrite(filename, frame)

    print("Saved: ", filename)

cap.release()
```

## video contact sheet generator

```python
import cv2
import numpy as np
timestamp = [0,14,17,9,15,2]
w,h = 320,180
colu = 2
cap = cv2.VideoCapture("running.mp4")
images = []
for sec in timestamp:
    cap.set(cv2.CAP_PROP_POS_MSEC, sec * 1000)
    ret, frame = cap.read()
    if ret:
        images.append(cv2.resize(frame,(w,h)))
cap.release()
rows = (len(images) + colu-1)//colu

sheet = np.zeros((rows*h,colu*w, 3), dtype = np.uint8)
for i ,image in enumerate(images):
    r,c = divmod(i, colu)
    sheet[
        r*h:(r+1)*h,
        c*w:(c+1)*w
        ]=image
cv2.imwrite("contact_sheet_3.jpg", sheet)
print("Saved")
```
