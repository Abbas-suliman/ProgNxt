```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

## 1.Geometric translation

```python
#Ex 1
image = cv2.imread('images/people.jpg')
cv2.imshow("Image display", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
h,w = image.shape[:2]
M = np.float32([[1,0,50] ,[0,1,30]])
shift = cv2.warpAffine(image, M, (w,h))

cv2.imshow("Image shiftted display", shift)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Ex 2
M1 = np.float32([[1,0,-50] ,[0,1,-30]])
shift = cv2.warpAffine(image, M1, (w,h))

cv2.imshow("Image shiftted display", shift)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Reflection

```python
rows, cols = image.shape[:2]
M = np.float32([[1,0,0] ,
                [0,-1,rows], # x axis
                [0,0,1]])
reflect = cv2.warpPerspective(image, M, (int(cols),int(rows)))

cv2.imshow("Image reflect display", reflect)
cv2.imwrite("Reflectedimage.jpg",reflect)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
rows, cols = image.shape[:2]
M = np.float32([[-1,0,cols] ,# y axis
                [0,1,0],
                [0,0,1]])
reflect1 = cv2.warpPerspective(image, M, (int(cols),int(rows)))

cv2.imshow("Image reflect display", reflect1)
cv2.imwrite("Reflectedimage.jpg",reflect1)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Rotation

```python
rows, cols = image.shape[:2]
M = np.float32([[1,0,0] ,
                [0,-1,rows],
                [0,0,1]])
M = cv2.getRotationMatrix2D((cols/2, rows/2),30, .6) # rotate => col and row div 2 around the center point, 30 its degree, 60 % reduce the size
rotate = cv2.warpAffine(image, M, (int(cols),int(rows)))

cv2.imshow("Image rotate display", rotate)
cv2.imwrite("rotatedimage.jpg",rotate)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
rows, cols = image.shape[:2]
M = np.float32([[1,0,0] ,
                [0,-1,rows],
                [0,0,1]])
M = cv2.getRotationMatrix2D((cols/2, rows/2),90, .6) # rotate => col and row div 2 around the center point, 30 its degree, 60 % reduce the size
rotate = cv2.warpAffine(image, M, (int(cols),int(rows)))

cv2.imshow("Image rotate display", rotate)
cv2.imwrite("rotatedimage.jpg",rotate)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Shearing

```python
rows, cols = image.shape[:2]
M = np.float32([[1,0.5,0] , # shearing x axis
                [0,1,0],
                [0,0,1]])

shear_X = cv2.warpPerspective(image, M, (int(cols*1.5),int(rows*1.5)))

cv2.imshow("Image shear_X display", shear_X)
cv2.imwrite("shear_X_image.jpg",shear_X)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
rows, cols = image.shape[:2]
M = np.float32([[1,0,0] , 
                [0.5,1,0],# shearing y axis
                [0,0,1]])

shear_Y = cv2.warpPerspective(image, M, (int(cols*1.5),int(rows*1.5)))

cv2.imshow("Image shear_Y display", shear_Y)
cv2.imwrite("shear_Y_image.jpg",shear_Y)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Fourier transformation

```python
image = cv2.imread('images/people.jpg')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

f = np.fft.fft2(gray)
f_shift  =np.fft.fftshift(f)
magnitude = 20* np.log(np.abs(f_shift)+1)

magnitude = cv2.normalize(magnitude, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

cv2.imshow("Transformation display", magnitude)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

GrabCut

```python
image = cv2.imread('images/messi5.jpg')
mask = np.zeros(image.shape[:2], np.uint8)
bgd = np.zeros((1,65), np.float64)
fgd = np.zeros((1,65), np.float64)

rect = (50,50, image.shape[1]-100, image.shape[0]-100)

cv2.grabCut(image, mask, rect, bgd, fgd, 5, cv2.GC_INIT_WITH_RECT)
m2 = np.where((mask==2)|(mask==0), 0,1).astype("uint8")

result=image*m2[:,:,None]

cv2.imshow("back display", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Hough lines

```python
image = cv2.imread('images/car.jpg')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) 
edges = cv2.Canny(gray, 50,150)
lines =cv2.HoughLines(edges, 1, np.pi/180, 120)
out = image.copy()

if lines is not None:
    for rho, theta in lines[:30, 0]:
        a,b = np.cos(theta), np.sin(theta)
        x0, y0 = a*rho, b*rho
        p1 = (int(x0+1000*(-b)), int(y0+1000*a))
        p2 = (int(x0-1000*(-b)), int(y0-1000*a))
        cv2.line(out, p1, p2, (0,255,0), 2)

cv2.imshow("line display", out)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
lines = cv2.HoughLinesP(edges, 1, np.pi/180, threshold = 50, minLineLength = 50, maxLineGap = 10)
print("segments:", 0 if lines is None else len(lines))

cv2.imshow("line display", out)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Hough circles

```python
image = cv2.imread('images/coins.jpg')
cv2.imshow("original display", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
image = cv2.imread('images/coins.jpg')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) 
gray = cv2.medianBlur(gray, 5)
circle =cv2.HoughCircles(gray, cv2.HOUGH_GRADIENT,dp= 1, minDist = 30, param1 =100, param2 = 30, minRadius = 10, maxRadius = 100)
out = image.copy()

if circle is not None:
    circle = np.uint16(np.around(circle))
    for x,y,r in circle[0]:
        cv2.circle(out, (x,y), r, (0,255,0),2)
    
cv2.imshow("Circle display", out)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
