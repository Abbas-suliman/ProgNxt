```python
np.sort(my_array)
```

```python
# text sorting
```

```python
my_array_1 = np.array(['red','pink','yellow','black','white', 'blue'])
my_array_1
```

```python
np.sort(my_array_1)
```

```python
# boolean
my_array_2 = np.array([False, True, True, True, False, False])
my_array_2
```

```python
np.sort(my_array_2)
```

```python
A = np.linspace(0,1,11)
A
```

```python
B = A==0.5
```

```python
B
```

```python
np.sort(B)
```

```python
# sorting in 2D
array_1 = np.random.randint(1,20, size=(4,6))
array_1
```

```python
np.sort(array_1)
```

```python
# descending order

my_array_1 = np.random.randint(1,20,10)
print(my_array_1)

sort_it = np.sort(my_array_1)
final_desc = np.flipud(sort_it)
print(final_desc)
```
