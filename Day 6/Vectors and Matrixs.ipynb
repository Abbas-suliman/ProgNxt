## multi dimen


```python
f = lambda m,n: n+10 * m
```

```python
A = np.fromfunction(f,(6,6), dtype = int)
A
```

```python
# third column
A[:,2]
```

```python
# third row
A[2,:]
```

```python
A
```

```python
# upper half diagonal block matrix
A[:3,:3]
```

```python
# lower left off diagonal block matrix
A[3:,:3]    
```

```python
A
```

```python
# every second element starting from 0,0
A[0::2,0::2]
```

```python
# every second element  and third element starting from 1,1
A[1::2,1::3]
```

```python
A
```

```python
# every second row element  and third column element starting from 0,0

A[0::2,0::3]
```

```python
A[::2,::2]
```
