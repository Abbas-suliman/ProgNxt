```python
import pandas as pd
```

```python
df1 = pd.DataFrame({'name':['vicky', 'micky'], 'age':[34,45]})
df2 = pd.DataFrame({'name':['steve', 'jobs'], 'age':[45,55]})
```

```python
df1
```

```python
df2
```

```python
# concate
pd.concat([df1,df2], axis = 1)
```

```python
df3 = pd.concat([df1,df2], axis = 0)
```

```python
df3
```

```python
df4 = pd.DataFrame({'name':['pinky', 'kamal'], 'age':[34,40], 'hobby':['reading','swimming']})
df4
```

```python
pd.concat([df3,df4], join = 'outer', ignore_index=True)
```

```python
pd.concat([df3,df4], join = 'inner', ignore_index=True)
```

merge

```python
size = pd.DataFrame({'color':['red','green','blue'],'size' :['M','S','XL']})
gender = pd.DataFrame({'color':['red','green','pink'],'gender' :['M','F','M']})
```

```python
size
```

```python
gender
```

```python
pd.merge(size,gender, how = 'outer')
```

```python
pd.merge(size,gender, how = 'inner')
```

```python
pd.merge(size,gender, how = 'left')
```

```python
pd.merge(size,gender, how = 'right')
```
