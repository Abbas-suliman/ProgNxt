```python
import pandas as pd
```

```python
dataset = pd.read_csv('gun-violence-data1.csv') 
dataset.head()
```

```python
dataset.drop(columns = ['incident_id','city_or_county','address'], inplace = True)
dataset.head()
```

```python
from datetime import datetime

#dataset['date'] = pd.to_datetime(dataset['date'], format = '%m-%d-%Y') # covert date into datetime formate
dataset['date'] = pd.to_datetime(dataset['date'], dayfirst=True) # covert date into datetime formate
dataset['year'] = dataset['date'].dt.year
dataset['month']= dataset['date'].dt.month
dataset['day']= dataset['date'].dt.day
dataset['day_of_week']= dataset['date'].dt.dayofweek

dataset['tot_victim'] = dataset['n_killed'] + dataset['n_injured']

dataset.head()
```

```python
dataset['day_of_week'] = dataset['day_of_week'].map({0:'Monday',1:'Tue',2:'Wed',3:'Thurs',4:'Friday',5:'Satur',6:'Sunday'})
dataset.head()
```

```python
dataset.isnull().sum()
```

Groupwise

```python
dataset['day_of_week'].value_counts()
```

```python
dataset.groupby('day_of_week')['tot_victim'].count()
```

```python
dataset.groupby('state')['tot_victim'].sum()
```

```python
import numpy as np
dataset.groupby('month')['tot_victim'].agg([np.mean, min, max])
```

Sorting

```python
dataset.groupby('state')['tot_victim'].sum().sort_values(ascending = False)
```

```python
dataset.groupby('month')['tot_victim'].count().sort_values(ascending = True)
```
