## 5. Rotation

```python
# Ex 1
# cv2.rotate => comman rotation
image = cv2.imread('image/opencv-logo.png')
rotate = cv2.rotate(
    image, cv2.ROTATE_90_CLOCKWISE)
cv2.imshow("Roated", rotate)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
image = cv2.imread('image/opencv-logo.png')
cv2.imshow("original", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Ex 2
# getRotationMatrix2D() + warpAffine()
image = cv2.imread('image/opencv-logo.png')

height , width = image.shape[:2] # row , col => y, x
center = (width//2, height//2)

matrix = cv2.getRotationMatrix2D(center, 160, 1.0)

rotated = cv2.warpAffine(image, matrix,(width,height) ) # x, y

cv2.imshow("Roated", rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 7. Split the Color Channel

```python


# B, G, R
image = cv2.imread('image/opencv-logo.png')

blue, green, red = cv2.split(image)

cv2.imshow("Blue", blue)
cv2.imshow("Green", green)
cv2.imshow("Red", red)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# B, G, R
import matplotlib.pyplot as plt
image = cv2.imread('image/opencv-logo.png')

blue, green, red = cv2.split(image)

plt.figure(figsize = (10,4))
plt.subplot(1,3,1)
plt.imshow(blue)

plt.figure(figsize = (10,4))
plt.subplot(1,3,2)
plt.imshow(green)

plt.figure(figsize = (10,4))
plt.subplot(1,3,3)
plt.imshow( red)
plt.show()
```

```python
# missed cmap => now its included here

import matplotlib.pyplot as plt
image = cv2.imread('image/opencv-logo.png')

blue, green, red = cv2.split(image)

plt.figure(figsize = (10,4))
plt.subplot(1,3,1)
plt.imshow(blue, cmap = 'gray')

plt.figure(figsize = (10,4))
plt.subplot(1,3,2)
plt.imshow(green, cmap = 'gray')

plt.figure(figsize = (10,4))
plt.subplot(1,3,3)
plt.imshow( red, cmap = 'gray')
plt.show()
```

## 8. Merging the color channel

```python
image = cv2.imread('image/opencv-logo.png')

blue, green, red = cv2.split(image)
# cv2.merge => combines in single array
merged = cv2.merge([blue, green, red])

cv2.imshow("Merged Output", merged)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
image = cv2.imread('image/opencv-logo.png')

blue, green, red = cv2.split(image)
# cv2.merge => combines in single array
merged = cv2.merge([red, green, blue])

cv2.imshow("Merged Output", merged)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 12. Blend 2 images

```python
image1 = cv2.imread('image/apple.png')
image2 = cv2.imread('image/gaussian-noise.png')

image2 = cv2.resize(image2, (image1.shape[1],image1.shape[0]) )

blend = cv2.addWeighted(image1, 0.5, image2, 0.5, 0)

cv2.imshow("blending 2 img", blend)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 14. image filtering and smoothing


```python
image1 = cv2.imread('image/shapes.jpg')

blurrr = cv2.GaussianBlur(image1,(7,7), 0)
cv2.imshow("Original", image1)
cv2.imshow("Blurred image", blurrr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Ex 2
image1 = cv2.imread('image/shapes.jpg')

blurrr = cv2.medianBlur(image1,7)
cv2.imshow("Original", image1)
cv2.imshow("Median Blurred image", blurrr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 15. Threshold

```python
# gray to binary
# OTSU = > it chooses its own thres automatically using histogram of the image
image1 = cv2.imread('image/apple.png')
hist = cv2.calcHist([image1], [0], None, [256], [0,256])
# display histo

plt.plot(hist)
plt.xlabel("Pixel intensity")
plt.ylabel("frequency")
plt.title("histo of an image")
plt.show()
```
