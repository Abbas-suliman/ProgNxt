Input: image.jpg 
Output: License plate

Pipeline:
Read image →
grayscale →
bilateralfilter →
edges detection→
contours detection →
Shape Approximation →
Aspect ration →
Plate number Extraction

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1
img = cv2.imread("car1.jpg")
```

```python
# 2
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
# 3
blur = cv2.bilateralFilter(gray,11,17,17)
```

```python
# 4
edges = cv2.Canny(blur, 30, 200)
```

```python
# 5
contours, _ = cv2.findContours(edges, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
contours = sorted(contours, key=cv2.contourArea, reverse=True)
```

```python
found = False # no plate has been detcted
```

```python
for c in contours:
    peri = cv2.arcLength(c, True)
    approx = cv2.approxPolyDP(c,0.02 * peri, True)
    if len(approx) ==4:
        x,y,w,h = cv2.boundingRect(approx)
        ratio = w/float(h)
        if 2.0 < ratio < 6.0 and w >80:
            plate = img[y:y+h, x:x+w]
            cv2.rectangle(img, (x,y), (x+w, y+h),(0,255,0),3)
            cv2.imwrite("Plate cropped.jpg", plate)
            found = True
            break
print("Plate region found", found)
# save it
cv2.imwrite("Plate_region.jpg", img)
cv2.imshow("Result", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
