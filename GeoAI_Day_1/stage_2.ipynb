```python
# Polygon
polygon = Polygon([
    (80.20,13.00),
    (80.30,13.00),
    (80.30,13.10),
    (80.20,13.10)
])

polygon.geom_type
```

```python
polygon.bounds
```

```python
polygon.boundary
```

```python
polygon.centroid
```

```python
polygon.area
```

```python
x, y = polygon.exterior.xy

# Draw the polygon
plt.fill(x, y, color='skyblue', alpha=0.5, label="Area")
plt.plot(x, y, color="blue", linewidth=2)

# Plot centroid
plt.scatter(
    polygon.centroid.x,
    polygon.centroid.y,
    color='red',
    label="Centroid"
)

# Label polygon vertices
for i, (x1, y1) in enumerate(polygon.exterior.coords[:-1]):
    plt.text(x1, y1, f"P{i+1}", fontsize=10)

# Title, labels, legend, grid
plt.title("Polygon with Centroid and Vertices")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.legend()
plt.grid(True)

plt.show()
```
